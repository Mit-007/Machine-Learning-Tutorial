```python
# Project Task:
# Build and compare KNN, Logistic Regression, and Naive Bayes
# classification models on the Iris dataset.
# Train the models using 50% of the data and test on 100% of the data.
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split ,GridSearchCV
from sklearn.metrics import accuracy_score,recall_score,precision_score
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv("/content/drive/MyDrive/Machine Learning/Supervised Learning/Classification /Iris.csv")
le = LabelEncoder()
data['Species'] = le.fit_transform(data['Species'])
data
```

```python
X = data.drop(["Species","Id"],axis=1)
y= data['Species']

print(X.shape)
print(y.shape)
```

```python
X_train , X_test , y_train , y_test = train_test_split(X,y,test_size=0.5,random_state =42)
```

```python
logi_model = LogisticRegression(max_iter=2000)
logi_model.fit(X_train , y_train)
```

```python
logistic_y_pred = logi_model.predict(X_test)
```

```python
gnb_model =  GaussianNB()
gnb_model.fit(X_train,y_train)
```

```python
gnb_y_pred = gnb_model.predict(X_test)
```

```python
pipeline = Pipeline([('scaler', StandardScaler()), ('knn_classifier',KNeighborsClassifier())])

param_grid = {"knn_classifier__n_neighbors" : [3, 5, 7, 9, 11, 13]}

classifierCV = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="accuracy"
)

classifierCV.fit(X_train,y_train)
```

```python
knn_y_pred = classifierCV.predict(X_test)
```

```python
print("-------------------logistic-----------------------")
print("Accuracy : ",accuracy_score(y_test,logistic_y_pred))

print("\n-------------------Naive bayes--------------------")
print("Accuracy : ",accuracy_score(y_test,gnb_y_pred))

print("\n-------------------KNN----------------------------")
print("Accuracy : ",accuracy_score(y_test,knn_y_pred))
```
