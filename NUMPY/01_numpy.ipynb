```python
size = 100_000_000

# make python list
py_list = list(range(size))


# make numpy array
np_arr = np.array(py_list)
```

```python
import time
```

```python
# performe time diffrence

start_time = time.time()
sq_list = [x**2 for x in py_list]
end_time = time.time()
print(f"total time for py_list : {end_time - start_time} seconds")

start_time = time.time()
sq_arr = np_arr ** 2
end_time = time.time()
print(f"total time for np_arr  : {end_time - start_time} seconds")
```

```python
# compare memory

import sys

print(f"pyhton list size in bytes : {sys.getsizeof(py_list) * len(py_list)}")

print(f"numpy array size in bytes : {np_arr.nbytes}")
```
