```python
import matplotlib.pyplot as plt
```

```python
expenses = ["Salaries","Rent","marketing","R&D","Miscellaneous"]
amount = [500,150,120,100,50]
```

```python
plt.pie(amount ,labels = expenses)
```

# customizations

```python
# autopct -> show the percentages of each slice

plt.pie(amount ,labels = expenses,autopct="%1.2f%%")
```

```python
# color  -> add color in slices
plt.pie(amount ,labels = expenses,autopct="%1.2f%%",colors=["red","green","blue","yellow","black"])
```

```python
# wedgeprops -define (edgecolor,linewidth,linestyle,alpha,fill,...)

plt.pie(amount,labels=expenses,autopct="%1.2f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":2,
            "linestyle":"--",
            "fill" : True,
            "alpha" : 0.5
        }
        )
```

```python
# explode ->highlights the slice with pulling out

# expenses = ["Salaries","Rent","marketing","R&D","Miscellaneous"]
# amount = [500,150,120,100,50]
explode = [0 , 0.15 ,0 , 0.2 , 0]

plt.pie(amount,labels=expenses,autopct="%1.2f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":2,
            "linestyle":"--",
            "fill" : True,
            "alpha" : 0.5
        },
        explode=explode
        )
```

```python
# startangle -> rotate the chart

explode = [0 , 0.15 ,0 , 0.2 , 0]

plt.pie(amount,labels=expenses,autopct="%1.2f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":2,
            "linestyle":"--",
            "fill" : True,
            "alpha" : 0.5
        },
        explode=explode,
        startangle=180
        )
```

```python
# shadow -> add 3D effect or makes chart circular

explode = [0 , 0.15 ,0 , 0.2 , 0]

plt.pie(amount,labels=expenses,autopct="%1.2f%%",
        wedgeprops={
            "edgecolor":"black",
            "linewidth":2,
            "linestyle":"--",
            "fill" : True
            },
        explode=explode,
        shadow=True
        )
```
