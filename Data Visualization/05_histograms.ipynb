```python
# multiple datasets

legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]
```

```python
plt.hist(legit_transactions,color="green",bins = 10,label="legit",edgecolor="black",alpha = 0.6)
plt.hist(fraud_transactions,color="red",bins = 10,label="fraud",edgecolor="black",alpha=0.4)

plt.title("Transaction amount Distribution")
plt.xlabel("Amount in $")
plt.ylabel("Frequency of trans")
plt.grid()
plt.legend()
```
