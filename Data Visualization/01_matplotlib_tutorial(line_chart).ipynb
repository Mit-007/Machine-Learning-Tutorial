```python
oscar_movie = [
    "Endgame",
    "Infinity War",
    "spider-man brand new world",
    "Iron-man 2",
    "Thor: Ragnarok",
    "Captain America: Civil War"
]
years = [2009, 2010, 2012, 2014, 2017, 2019]
oscar_revenue = [1050, 250, 2354, 254, 2548, 1000] #in $M
```

```python
plt.plot(years,oscar_revenue)
```

```python

plt.plot(years,oscar_revenue)

# title()

plt.title("Oscar Movie revenue each year")

# xlabel()

plt.xlabel("Year")

# ylabel()

plt.ylabel("Revenue (in $M)")
```

```python
Non_oscar_movie = [
    "FF1",
    "pirates of the caribbean",
    "batman",
    "The Dark Knight",
    "thr godFather",
    "The Lord of the Rings: The Return of the King"
]
non_oscar_revenue = [2050, 1250, 1354, 2054, 1548, 1058] #in $M
```

#formating line in graph

```python
plt.plot(years,oscar_revenue ,label = "Oscar winner")
plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

plt.title("revenue of oscar vs non oscar movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# legend()

plt.legend()
```

# Methods And Attributes :

```python
# style.available  : list of available style

plt.plot(years,oscar_revenue ,label = "Oscar winner")
plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

plt.title("revenue of oscar vs non oscar movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# legend()

plt.legend()

plt.style.available
```

```python
# style.use : chnage a style

plt.plot(years,oscar_revenue ,label = "Oscar winner")
plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

plt.title("revenue of oscar vs non oscar movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# legend()

plt.legend()
plt.style.use("dark_background")
```

```python
# show() : show plot

plt.plot(years,oscar_revenue ,label = "Oscar winner")
plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

plt.title("revenue of oscar vs non oscar movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# legend()

plt.legend()

plt.style.available
plt.show()
```

```python
# plt.grid() : add grid in graph

plt.plot(years,oscar_revenue ,label = "Oscar winner")
plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

plt.title("revenue of oscar vs non oscar movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

# legend()

plt.legend()
plt.style.use("default")
plt.grid()
```

```python
# xkcd : print in cartoon style

with plt.xkcd():
  plt.plot(years,oscar_revenue ,label = "Oscar winner")
  plt.plot(years,non_oscar_revenue , label = "Non-oscar winner")

  plt.title("revenue of oscar vs non oscar movie")
  plt.xlabel("Year")
  plt.ylabel("Revenue (in $M)")

  plt.legend()
```

```python
# tight_layout  : avoid a some overlapping in labeling and make space between it (not remove comaplete overlapping)

movie = [
    "Endgame",
    "Infinity War",
    "spider-man ",
    "Iron-man 2",
    "Thor: Ragnarok",
    "Captain America"
]
years = [2009, 2010, 2012, 2014, 2017, 2019]
revenue = [1050, 250, 2354, 254, 2548, 1000] #in $M

plt.plot(movie , revenue, label = "Non-oscar winner")

plt.title("revenue of movie")
plt.xlabel("movie")
plt.ylabel("Revenue (in $M)")

plt.tight_layout()
```
