```python
import matplotlib.pyplot as plt
```

```python
# plot methods() :

# 1. plt.plot(x, y) - Line plot

# 2. plt.bar(x, height) - Vertical bar chart

# 3. plt.barh(y, width) - Horizontal bar chart

# 4. plt.scatter(x, y) - Scatter plot

# 5. plt.pie(data) - Pie chart

# 6. plt.hist(data, bins) - Histogram

# 7. plt.boxplot(data) - Box plot
```

```python
years = [2008, 2010, 2012, 2014, 2016, 2018]
revenue = [1050, 250, 2354, 254, 2548, 1000]
```

```python
# vertical bar chart
plt.bar(years, revenue)

plt.title("revenue for each year")
plt.xlabel("years")
plt.ylabel("revenue in $M")
```

```python
#  horizontol bar chart

plt.barh(years, revenue ,color ="green")

plt.title("revenue for each year")
plt.xlabel("years")
plt.ylabel("revenue in $M")
```

```python
# add label of in bar

# plt.text(x-axis position , y-axix position , text , alignment)

plt.bar(years, revenue)

plt.title("revenue for each year")
plt.xlabel("years")
plt.ylabel("revenue in $M")

plt.text(years[0] , revenue[0] + 100 , "ABC" , ha = "center")
```

```python
plt.bar(years, revenue)

plt.title("revenue for each year")
plt.xlabel("years")
plt.ylabel("revenue in $M")


for i in range(len(years)):
  plt.text(years[i] , revenue[i] + 10 , revenue[i] , ha = "center")

# set limit for graph
plt.xlim(2000,2018)
plt.ylim(0,max(revenue)+100)
```

```python
oscar_movie = [
    "Endgame",
    "Infinity War",
    "spider-man brand new world",
    "Iron-man 2",
    "Thor: Ragnarok",
    "Captain America: Civil War"
]
Non_oscar_movie = [
    "FF1",
    "pirates of the caribbean",
    "batman",
    "The Dark Knight",
    "thr godFather",
    "The Lord of the Rings: The Return of the King"
]
hindi_movie = [
    "animal",
    "Dhhurndher",
    "dangal",
    "bahubali 2",
    "RRR",
    "Pushpa"
]
oscar_revenue = [1050, 250, 2354, 254, 2548, 1000]
non_oscar_revenue = [2050, 1250, 1354, 2054, 1548, 1058]
hindi_revenue = [200, 3250, 1954, 205, 154, 1505]
years = [2008, 2010, 2012, 2014, 2016, 2018]
```

```python
# overlapping bars
plt.bar(years,oscar_revenue)
plt.bar(years,non_oscar_revenue)

plt.title("Oscar vs Non-oscar")
plt.xlabel("years")
plt.ylabel("revenue in $M")
```

```python
plt.bar(years,oscar_revenue)
plt.bar(years,non_oscar_revenue)

plt.title("Oscar vs Non-oscar")
plt.xlabel("years")
plt.ylabel("revenue in $M")
```

```python
import numpy as np
```

```python
x = np.arange(len(years))

width = 0.4
plt.bar(x-width/2,oscar_revenue,width,label = "Oscar")
plt.bar(x+width/2,non_oscar_revenue,width,label = "Non-Oscar")

plt.title("Oscar vs Non-oscar")
plt.xlabel("years")
plt.ylabel("revenue in $M")
plt.legend()
```

```python
x = np.arange(len(years))

width = 0.3
plt.bar(x-width,oscar_revenue,width,label = "Oscar")
plt.bar(x,non_oscar_revenue,width,label = "Non-Oscar")
plt.bar(x+width,hindi_revenue,width,label = "Hindi Movie")

plt.title("Oscar vs Non-oscar")
plt.xlabel("years")
plt.ylabel("revenue in $M")
plt.legend()

plt.xticks(x,years)
```
