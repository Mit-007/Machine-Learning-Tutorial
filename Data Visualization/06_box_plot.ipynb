```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [7 , 8 , 5 , 6 , 9 ,4 , 10 ,12 ,15]
```

```python
plt.boxplot(data)
plt.grid()
```

```python
data = [7 , 8 , 5 , 6 , 9 ,4 , 10 ,12 ,15 ,50]

plt.boxplot(data)
plt.grid()
```

# perform operation on box plot

```python
# multiple dataset

group1= np.random.normal(100, 10, 200)
group2= np.random.normal(80, 30, 200)
group3= np.random.normal(90, 20, 200)

plt.boxplot([group1,group2,group3] , tick_labels=["group1","group2","group3"])
plt.grid()
```

```python
# horizonatl box plot
data = [7 , 8 , 5 , 6 , 9 ,4 , 10 ,12 ,15 ,50]

plt.boxplot(data ,vert=False)
plt.grid()
```

```python
# show mean

group1= np.random.normal(100, 10, 200)
group2= np.random.normal(80, 30, 200)
group3= np.random.normal(90, 20, 200)

plt.boxplot([group1,group2,group3] , tick_labels=["group1","group2","group3"] ,showmeans=True)
plt.grid()
```

```python
# change a whisker length

data = [7 , 8 , 5 , 6 , 9 ,4 , 10 ,12 ,15]

plt.boxplot(data ,whis=1.0)
plt.grid()
```
