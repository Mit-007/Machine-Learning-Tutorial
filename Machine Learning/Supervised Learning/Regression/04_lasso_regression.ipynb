```python
import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
insurance_data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Supervised Learning/00_starting_ML/insurance.csv")
```

```python
insurance_data.head()
```

```python
X2 = insurance_data.drop(columns = ["charges"])
y2= insurance_data["charges"]

X2 = pd.get_dummies(X2,columns=["region"],drop_first=False)

X2["sex"] = X2["sex"].map({"female":1 ,"male":0})
X2["smoker"] = X2["smoker"].map({"yes":1 ,"no":0})

X2["age_smoker"] = X2["age"] * X2["smoker"]
X2["bmi_smoker"] = X2["bmi"] * X2["smoker"]

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, random_state=42)

X2.head()
```

```python
lasso_model = Lasso(alpha=0.5)
lasso_model.fit(X2_train, y2_train)
```

```python
# predict value

y2_pred = lasso_model.predict(X2_test)
```

```python
mse = mean_squared_error(y2_test, y2_pred)
print("Mean Squared Error:", mse)
```

# LASSO-CV

### find best alpha for model

```python
from sklearn.linear_model import LassoCV

a = [0.001 ,0.1 ,1 , 2, 5,10,20,30,40,50,100]

lasso_cv_model = LassoCV(
    alphas = a,
    cv = 5,
    max_iter = 1000,
    random_state = 42
)

lasso_cv_model.fit(X2_train,y2_train)

print("best aplha for lasso : ", lasso_cv_model.alpha_)

y2_pred = lasso_cv_model.predict(X2_test)

mse = mean_squared_error(y2_test, y2_pred)
print("Mean Squared Error:", mse)
```
