# large example for better understanding
## non linear dataset

```python
from sklearn.datasets import make_circles

X, _ = make_circles(n_samples = 500, factor=.5, noise=.03, random_state=42)
```

```python
plt.scatter(x=X[:,0],y=X[:,1],c="red")
plt.show()
```

```python
dbscan = DBSCAN(eps = 0.11,min_samples=5)
cluster = dbscan.fit_predict(X)
```

```python
plt.scatter(x=X[:,0],y=X[:,1],c=cluster,cmap='viridis',marker='o')
plt.show()
```

## part-2 Non-linear dataset (compare with Kmeans Algo)

```python
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN,KMeans
import matplotlib.pyplot as plt

X, _ = make_moons(n_samples = 500, noise=.03, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
scatters = plt.scatter(x=X[:,0],y=X[:,1],c="red")
plt.show()
```

```python
dbscan = DBSCAN(eps = 0.2,min_samples=5)
dbscan_cluster = dbscan.fit_predict(X_scaled)
```

```python
kmeans = KMeans(n_clusters=2)
kmean_cluster = kmeans.fit_predict(X_scaled)
```

```python
scatters = plt.scatter(x=X_scaled[:,0],y=X_scaled[:,1],c=kmean_cluster,cmap='viridis',marker='o')
plt.title("KMean Clustering Result")
plt.show()
```

```python
scatters = plt.scatter(x=X_scaled[:,0],y=X_scaled[:,1],c=dbscan_cluster,cmap='viridis',marker='o')
plt.title("DBSCAN Clustering Result")
plt.show()
```

# Apply DBSCAN on iris Dataset

```python
from sklearn.datasets import load_iris
from sklearn.cluster import DBSCAN
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],hue=y)
```

```python
dbscan = DBSCAN(eps=0.8,min_samples=5)
cluster = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],hue=cluster)
```
