```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
iris = load_iris()
```

```python
X = iris.data
```

```python
y = iris.target
```

```python
X.shape
```

```python
y
```

```python
sns.scatterplot(x=X[:,0],y=X[:,2],c=y)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
wcss = []

for i in range(1,10):
  kmeans = KMeans(n_clusters=i,random_state=42)
  kmeans.fit(X_scaled)
  wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,10), y=wcss, marker='o')
```

```python
kmeans = KMeans(n_clusters=3)

lables = kmeans.fit_predict(X_scaled)

sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c=lables)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,2],marker="x",c="red",s=200)
```

# PCA - Dimensionality reduction

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2) # convert 4 feature into 2

pca_data = pca.fit_transform(X_scaled)
```

```python
kmenas1 = KMeans(n_clusters=3,random_state=42)

lables = kmenas1.fit_predict(pca_data)
```

```python
sns.scatterplot(x=pca_data[:,0],y=pca_data[:,1],c=lables)
sns.scatterplot(x=kmenas1.cluster_centers_[:,0],y=kmenas1.cluster_centers_[:,1],marker="x",c="red",s=200  )
```
