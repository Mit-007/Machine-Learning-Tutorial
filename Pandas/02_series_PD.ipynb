# Series Properties

```python
# homogenous (all data are same type)
ser3= pd.Series([1,2,3,4,5,6,"w"] , index = ["a","b","c","x","y","z",2])
ser3

# dtype is now object
```

```python
# vectorization
ser4 = pd.Series([1,2,3,4,5,6])
ser5 = pd.Series([1,2,3,4,5,6])

print(ser4 + ser5)
```

```python
# handle missing values with NAN

# have muatble value , immutable size
ser6 = pd.Series([1,2,3,4,5,6])

ser6[4] = 100

ser6
```
