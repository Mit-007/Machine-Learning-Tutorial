# filtering of data

```python
# filter all city , those have aqi > 100

df[df["aqi"] > 100]
```

```python
# those city have aqi > 100 and temperature > 30
df[ (df["aqi"] > 100) & (df["temperature"] > 30)]
```

```python
# chaining of filter
# first select high aqi city , then print city + aqi columns only
df[df["aqi"] > 100][["city" , "aqi"]]
```

```python
# --> filtering using query Method
#  query return copy not view

df.query("aqi > 100")
df.query("aqi > 100 & temperature > 30")

aqi_limit = 100
temp_limit = 30
df.query("aqi > @aqi_limit & temperature > @temp_limit ")
```
