# cleaning data :   
## 1. handle missing vale

```python
# find which cell have null value (True indicate null)
df.isnull()
df.isna()
```

```python
# find total null value per columns

df.isnull().sum()
```

```python
# drop data

# drop rwo
df.dropna()

# drop column
df.dropna(axis = 1)
```

```python
# fill a null value with , given value

df.fillna(0)

age_mean = df["age"].mean()

df["age"].fillna(age_mean)

df.ffill() #forword fill , set value of upper cell value

df.bfill() #backword fill , set value of lower cell value
```

## 2.handle duplicates data

```python
df.duplicated()

df['country'].duplicated() #check for perticular columns

df[['age','country']].duplicated()
```

```python
# drop duplicates

df.drop_duplicates()
# df.drop_duplicates(inplace=True) #inplace means changes done in give origianl data set
```

## 3. Handle data type & date-time

```python
# data types

df.dtypes
```

```python
df2 = df.copy()

df2_age_mean = df2['age'].mean()
df2["age"] = df2["age"].fillna(df2_age_mean)
df2["age"] = df2["age"].astype("int64")

df2.dtypes
```

```python
# convert any data type into date-time

date_str = pd.Series([pd.to_datetime("2026-7-13")])
print(date_str.dtypes)
```
