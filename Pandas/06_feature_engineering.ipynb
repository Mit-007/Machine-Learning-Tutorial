# Transforming Data ( Feature Engineering )

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Pandas/employee_data.csv")

df
```

###  apply()

```python


df2 = df.copy()

df2['text'] = df2['Salary'].apply(lambda x : "20%" if x >=50000 else "10%")
```

```python
df2
```

### map()

```python
df3 = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Pandas/raw_data.csv")
df3
```

```python
df4 = df3.copy()
```

```python
map_gender = {"Male" : "M", "Female" : "F","Unknown" : "U"}

df4['gender'] = df4['gender'].map(map_gender)
```

```python
df4
```

### assign()

```python
df4.assign(new_income = df4['income'] * 1.1)
```

### replace()

```python
df4['country'] = df4['country'].replace("USA","United State")
```

### rename()

```python
# rename all columns name

df3.columns = ["Id","Name","Age","Country","Gender","Income"]

df3

# change one column

df3.rename(columns = {"Income" : "Salary"})

# chnage label

df3.rename(index = {1 : "One"})
```
