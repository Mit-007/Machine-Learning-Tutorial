```python
import pandas as pd
```

```python
df_coustomers = pd.DataFrame({
    "coustomer_id" : [1,2,3,4],
    'name' : ["mit","kavy","kishan","dev"]
})

df_orders = pd.DataFrame({
    'order_id' : [101,102,103,104],
    'coustomer_id' : [2,1,3,5],
    'amount' : [250 , 120 ,300 ,180]
})
```

```python
df_coustomers
```

```python
df_orders
```

```python
# inner join
pd.merge(df_coustomers,df_orders,on='coustomer_id')
```

```python
# left join
pd.merge(df_coustomers,df_orders,on="coustomer_id",how="left")
```

```python
# Right join
pd.merge(df_coustomers,df_orders,on="coustomer_id",how="right")
```

```python
# outer join
pd.merge(df_coustomers,df_orders,on="coustomer_id",how="outer")
```
